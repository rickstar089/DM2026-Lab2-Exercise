# **Project Report**

The objective of this project was to develop a robust classification model for identifying six emotions in Twitter posts: anger, disgust, fear, joy, sadness, and surprise. The development process progressed from classical machine learning to pretrained Transformers and ultimately to a hybrid ensemble combining complementary classification approaches.

### 1.1 Preprocessing Steps

The dataset consisted of 3,360 labeled training posts and 1,440 unlabeled competition test posts. The initial data analysis investigated class distributions, missing values, duplicate texts, conflicting labels, unusual text lengths, and potential outliers.

To ensure consistent and leakage-aware evaluation, the labeled dataset was divided into a development set of 2,691 examples and a reserved holdout set of 669 examples. Five-fold stratified group cross-validation was applied to the development set, ensuring that identical and identified near-duplicate texts were assigned to the same group. These validation folds remained fixed throughout model development.

Several preprocessing techniques were evaluated individually and in combination:

- Lowercasing and whitespace normalization.
- HTML entity decoding and Unicode normalization.
- Normalization of URLs and user mentions.
- Hashtag processing while retaining their textual content.
- Emoji, punctuation, and repetition handling.
- Alternative treatments of duplicates, conflicting labels, and outliers.

The initial TF-IDF and LinearSVC baseline achieved a mean cross-validation Macro-F1 of 0.4509. Lowercasing improved this score to 0.4651, while the selected combined preprocessing pipeline, P06, achieved 0.4678.

Aggressive text cleaning was avoided because capitalization, punctuation, emojis, and repeated characters may contain valuable emotional information. Experiments involving the removal or reweighting of potentially problematic training examples did not demonstrate a sufficiently convincing improvement. Consequently, all original training examples and labels were retained.

Different preprocessing strategies were used depending on the feature representation. Classical word-based features used the selected normalization pipeline, while character and structural features preserved additional surface-level information. Pretrained Transformer models primarily relied on their native tokenizers.

### 1.2 Feature Engineering Steps

The feature engineering process investigated three complementary categories of text representations.

**Lexical features**

Word-level TF-IDF unigrams and bigrams were used to capture individual words and short expressions associated with emotions. Character-level TF-IDF n-grams of lengths three to six were introduced to capture spelling variations, word fragments, informal language, and other patterns frequently found in social media posts.

Chi-square feature selection was applied to retain the 20,000 most relevant features. Vocabulary construction, feature scaling, and feature selection were performed exclusively on the respective training folds.

**Structural features**

A total of 21 additional structural features were extracted, including:

- Text length, word count, and average word length.
- Capitalization and uppercase word frequency.
- Exclamation marks, question marks, and punctuation patterns.
- Character elongation and word repetition.
- Emojis, emoticons, hashtags, mentions, and URLs.

These features were scaled using MaxAbsScaler and combined with the lexical representations. Their purpose was to capture emotional intensity and writing style beyond conventional word frequencies.

**Semantic features**

To overcome the limitations of sparse lexical representations, pretrained Transformer encoders were introduced. MiniLM and the CLIP text encoder provided contextual representations capable of capturing semantic relationships between words and sentences.

Additional experiments investigated Twitter-RoBERTa, BERTweet, DeBERTa, ELECTRA, ModernBERT, RoBERTa, and BGE. Different pooling strategies, frozen embeddings, partial fine-tuning, and full fine-tuning were compared.

The most important subsequent improvement came from emotion-specific pretrained models. Emotion RoBERTa Large and Emotion DistilRoBERTa provided representations already adapted to emotion recognition, making them particularly suitable for the classification task.

Combining lexical, structural, and semantic information allowed the models to capture different aspects of emotional expression rather than relying on a single feature representation.
### 1.3 Explanation of Your Model

**Classical machine learning baseline**

The first competitive model, C1_structure, combined word-level TF-IDF, character-level TF-IDF, structural features, and chi-square feature selection with a LinearSVC classifier.

The selected classifier used a regularization parameter of C = 0.3 and balanced class weights to address class imbalance. This model achieved a five-fold Macro-F1 of 0.5199 and a public Kaggle score of 0.5683.

Although this represented a substantial improvement over the initial baseline, its ability to recognize context-dependent emotions remained limited.

**Transformer ensembles**

The next development stage investigated pretrained Transformer models. One successful approach, E08, combined five fine-tuned models:

- Three MiniLM configurations.
- Two CLIP text encoder configurations.

Probability averaging was used to combine their predictions, assigning equal total weight to the two encoder families. E08 achieved a five-fold Macro-F1 of 0.5940 and a public Kaggle score of 0.6111.

The subsequent P6E02 model extended the semantic ensemble approach with specialized binary classifiers for frequently confused emotion pairs, including anger/disgust, anger/sadness, joy/sadness, and joy/surprise.

These specialists combined semantic embeddings with lexical features and were activated only when the underlying model was uncertain between two relevant emotions. P6E02 improved the five-fold Macro-F1 to 0.6052.

**Emotion-specific transfer learning**

Further experiments with general-purpose pretrained Transformers, alternative learning rates, class weighting, LoRA, and full fine-tuning produced limited improvements.

This motivated a change in strategy toward models already pretrained for emotion classification:

- `j-hartmann/emotion-english-roberta-large`
- `j-hartmann/emotion-english-distilroberta-base`

Both models originally predicted seven emotions, including neutral. Their pretrained classification heads were preserved, while the neutral output was excluded before applying softmax over the six competition classes.

Several adaptation strategies were evaluated, including head-only training, partial fine-tuning, full fine-tuning, and LoRA.

The strongest individual model, EM06, used Emotion RoBERTa Large with a frozen encoder and a trainable pretrained classification head. It achieved a five-fold Macro-F1 of 0.6208.

Emotion DistilRoBERTa with head-only adaptation, EM03, achieved 0.6105. Although its individual performance was lower, its predictions provided complementary information for the ensemble.

**Final ensemble**

The final model combined four complementary classification systems using weighted probability averaging:

| Model | Contribution | Weight |
|---|---|---:|
| EM06 | Emotion RoBERTa Large | 48% |
| E08 | MiniLM and CLIP ensemble | 16% |
| P6E02 | Hybrid model with emotion-pair specialists | 16% |
| EM03 | Emotion DistilRoBERTa | 20% |

For each test post, the six-class probability vectors were combined using these fixed weights. The emotion with the highest combined probability was selected as the final prediction.

For EM06 and EM03, test probabilities were averaged across the five saved fold-specific models before ensemble aggregation.

The four-component ensemble improved upon the previous three-component ensemble on four out of five validation folds, correcting 42 previously incorrect predictions while introducing 31 new errors.

**Experimental results**

| Development stage | Five-fold Macro-F1 |
|---|---:|
| Initial TF-IDF baseline | 0.4509 |
| Optimized preprocessing | 0.4678 |
| Classical model C1_structure | 0.5199 |
| MiniLM/CLIP ensemble E08 | 0.5940 |
| Emotion-pair specialists P6E02 | 0.6052 |
| Emotion RoBERTa Large EM06 | 0.6208 |
| Three-component ensemble | 0.6283 |
| **Final four-component ensemble** | **0.6316** |

The final ensemble achieved a pooled out-of-fold Macro-F1 of 0.6329 and a private Kaggle leaderboard score of **0.6433**, reaching **rank 37** and exceeding the original target score of 0.64.

The results demonstrate the benefit of combining task-specific pretrained knowledge with complementary lexical and semantic classification approaches. Nevertheless, repeated experimentation on the same development folds introduces potential model-selection bias, and local cross-validation scores should not be interpreted as guaranteed competition performance.

```python
# Normalize tweets and construct group-aware development and holdout splits.
import os, json, csv, hashlib, html, re, unicodedata
from pathlib import Path
from difflib import SequenceMatcher
import numpy as np
import pandas as pd
import regex
from sklearn.feature_extraction.text import HashingVectorizer
from sklearn.model_selection import StratifiedGroupKFold

LABELS = ("anger", "disgust", "fear", "joy", "sadness", "surprise")
SEED = 42
DATA_DIR = Path(os.environ.get("DM_DATA_DIR", "competition_data"))
MODEL_DIR = Path(os.environ.get("DM_MODEL_DIR", "competition_models"))
# Original competition files (not the Master notebook's SemEval dataset).
def require(condition, message):
    if not condition:
        raise ValueError(message)

def sha(path):
    return hashlib.sha256(Path(path).read_bytes()).hexdigest()

MENTION = re.compile(r"@<username>|(?<!\w)@[A-Za-z0-9_]+|(?<!\w)u/[A-Za-z0-9_-]+", re.I)
# Match both literal web links and the dataset's already anonymized ##URL### forms.
URL = re.compile(r"https?://[^\s<>]+|www\.[^\s<>]+|(?:\#{2,}URL)+\#{2,}|(?<!\w)URL(?!\w)", re.I)
HASHTAG = re.compile(r"(?<![\w#])#([\w]+)")
GRAPHEMES = regex.compile(r"\X")
EMOJI = regex.compile(r"\p{Extended_Pictographic}|\p{Regional_Indicator}|[0-9#*]\uFE0F?\u20E3")
PUNCT = re.compile(r"[^\w\s]+", re.UNICODE)


# Use a case-insensitive fingerprint to identify duplicate texts.
def canonical(s):
    """Conservative identity fingerprint for grouping; never a default model preprocessor."""
    return " ".join(unicodedata.normalize("NFKC", html.unescape(s)).casefold().split())

def whitespace(s):
    return " ".join(s.split())

# Decode HTML entities and normalize Unicode and whitespace.
def minimal(s):
    return whitespace(unicodedata.normalize("NFKC", html.unescape(s)))

def normalize_url(s):
    return URL.sub(" URL_TOKEN ", s)

def normalize_mention(s):
    return MENTION.sub(" MENTION_TOKEN ", s)

def normalize_hashtag(s):
    # Preserve hashtag word and its case; add an explicit shared marker, no segmentation.
    return HASHTAG.sub(lambda m: " HASHTAG_TOKEN " + m.group(1) + " ", s)

def emoji_graphemes(s):
    return [g for g in GRAPHEMES.findall(s) if EMOJI.search(g)]

# Use stronger normalization for word features and minimal cleaning for style features.
def prepare_text(text, recipe):
    if recipe == "minimal":
        return minimal(text)
    if recipe == "p06":
        for transform in (minimal, normalize_url, normalize_mention, normalize_hashtag):
            text = transform(text)
        return text
    if recipe in {"raw", "lower"}:
        return text
    raise ValueError(recipe)

# Lowercasing is performed by the lexical vectorizer, not by P06 itself.
def near_pairs(data):
    """Candidate search with fixed hash features (no learned vocabulary/IDF).

    Defined screening, not exhaustive semantic deduplication: normalized texts >=20
    characters; hash char 3..5 cosine >=.85; difflib ratio >=.90. Exact canonical
    equality is tabulated separately. Hash collisions / screening can miss pairs.
    """
    texts = data.text.map(canonical).tolist()
    # Screen similar character n-grams before checking sequence similarity.
    hv = HashingVectorizer(analyzer="char", ngram_range=(3, 5), n_features=2**20,
                          alternate_sign=False, binary=True, lowercase=False, norm="l2")
    mat = hv.transform(texts)
    pairs = []
    for start in range(0, len(data), 200):
        sims = (mat[start:start+200] @ mat.T).tocoo()
        for a, b, sim in zip(sims.row, sims.col, sims.data):
            i, j = start + int(a), int(b)
            if j <= i or sim < .85 or texts[i] == texts[j]: continue
            if min(len(texts[i]), len(texts[j])) < 20: continue
            if min(len(texts[i]), len(texts[j])) / max(len(texts[i]), len(texts[j])) < .82: continue
            ratio = SequenceMatcher(None, texts[i], texts[j], autojunk=False).ratio()
            if ratio >= .90:
                ra, rb = data.iloc[i], data.iloc[j]
                pairs.append({"post_id_a": ra.post_id, "set_a": ra['set'], "label_a": ra.label,
                              "text_a": ra.text, "post_id_b": rb.post_id, "set_b": rb['set'],
                              "label_b": rb.label, "text_b": rb.text,
                              "hash_char_cosine": float(sim), "sequence_ratio": ratio})
    return pd.DataFrame(pairs, columns=['post_id_a','set_a','label_a','text_a','post_id_b','set_b',
                                      'label_b','text_b','hash_char_cosine','sequence_ratio'])

# Merge connected duplicate pairs into the same group.
class UnionFind:
    def __init__(self, ids): self.p={i:i for i in ids}
    def find(self,x):
        if self.p[x]!=x: self.p[x]=self.find(self.p[x])
        return self.p[x]
    def union(self,a,b):
        a,b=self.find(a),self.find(b)
        self.p[max(a,b)]=min(a,b)

# Join text, partition and label tables by post ID; preserve submission order.
def load_competition(data_dir=DATA_DIR):
    posts = pd.DataFrame(json.loads((data_dir / "final_posts.json").read_text()))
    ident = pd.read_csv(data_dir / "data_identification.csv", dtype=str)
    labels = pd.read_csv(data_dir / "emotion.csv", dtype=str)
    sample = pd.read_csv(data_dir / "sampleSubmission.csv", dtype=str)
    data = posts.merge(ident, on="post_id", validate="one_to_one").merge(
        labels, on="post_id", how="left", validate="one_to_one")
    train_ids = ident.loc[ident["set"] == "train", "post_id"].tolist()
    test_ids = sample["post_id"].tolist()
    require(len(set(train_ids)) == len(train_ids) == 3360, "Training IDs")
    require(len(set(test_ids)) == len(test_ids) == 1440, "Test IDs")
    require(set(train_ids).isdisjoint(test_ids), "Overlapping partitions")
    require(set(test_ids) == set(ident.loc[ident["set"] == "test", "post_id"]), "Test partition")
    require(set(train_ids) == set(labels.post_id) and len(labels) == 3360, "Released labels")
    require(set(labels.label) == set(LABELS), "Invalid labels")
    require(len(data) == len(posts) == 4800, "Corpus size")
    indexed = data.set_index("post_id")
    train = indexed.loc[train_ids].reset_index().to_dict("records")
    test = indexed.loc[test_ids].reset_index()[["post_id", "text"]].to_dict("records")
    return data, train, test

def frozen_development(data, assignments_path=DATA_DIR / "assignments.csv"):
    # Keep the validation assignments fixed across experiments.
    train = data[data["set"] == "train"].copy().reset_index(drop=True)
    if assignments_path.exists():
        assignments = pd.read_csv(assignments_path, dtype={"post_id": str, "group_id": str})
    else:
        uf = UnionFind(train.post_id)
        for _, group in train.assign(key=train.text.map(canonical)).groupby("key"):
            for pid in group.post_id.iloc[1:]:
                uf.union(group.post_id.iloc[0], pid)
        near = near_pairs(data)
        near = near[(near.set_a == "train") & (near.set_b == "train")]
        for row in near.itertuples():
            uf.union(row.post_id_a, row.post_id_b)
        # Reserve one grouped split as a holdout and split the remainder into five folds.
        groups = train.post_id.map(uf.find)
        outer = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)
        dev, hold = next(outer.split(train.text, train.label, groups))
        assignments = train[["post_id", "label"]].copy()
        assignments["group_id"] = groups
        assignments["partition"] = "development"
        assignments.loc[hold, "partition"] = "holdout"
        assignments["cv_fold"] = -1
        inner = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)
        for fold, (_, valid) in enumerate(inner.split(
                train.text.iloc[dev], train.label.iloc[dev], groups.iloc[dev])):
            assignments.loc[dev[valid], "cv_fold"] = fold
    expected = "a09aec1349db78b0be4a2d68821edd977d5ed8814236730db5301974d1633bd6"
    # A hash mismatch prevents silently training on different validation folds.
    actual = hashlib.sha256(assignments.to_csv(index=False).encode()).hexdigest()
    require(actual == expected, "Frozen split differs; supply original assignments.csv")
    require(assignments.groupby("group_id").partition.nunique().max() == 1, "Group leakage")
    development = assignments[assignments.partition == "development"]
    require(len(development) == 2691 and len(assignments) == 3360, "Partition sizes")
    require(development.groupby("group_id").cv_fold.nunique().max() == 1, "Fold leakage")
    texts = train.set_index("post_id").text
    rows = development.assign(text=development.post_id.map(texts)).to_dict("records")
    return rows

# Retain all original training examples and labels.

```

## 3. Model Implementation Steps

```python
# classifiers Training and combining their six-class probability predictions.

def sync(device):
    if device.type == "mps":
        torch.mps.synchronize()

# Convert pooled encoder representations into six emotion logits.
class Classifier(nn.Module):
    def __init__(self, encoder, dropout, pooling):
        super().__init__()
        self.encoder = encoder
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(encoder.config.hidden_size, len(LABELS))
        self.pooling = pooling

    def forward(self, tokens):
        outputs = self.encoder(**tokens)
        hidden = outputs.last_hidden_state
        # Mask padding when averaging token representations; CLIP uses EOS pooling.
        if self.pooling == "mean":
            mask = tokens["attention_mask"].unsqueeze(-1).to(hidden.dtype)
            representation = (hidden * mask).sum(dim=1) / mask.sum(dim=1).clamp_min(1)
        elif self.pooling == "eos":
            representation = outputs.pooler_output
        else:
            representation = hidden[:, 0]
        return self.classifier(self.dropout(representation))

# Increase the loss contribution of rarer classes when weighting is enabled.
def loss_weights(y, cfg, device):
    if cfg["class_weight_power"] == 0:
        return None
    counts = np.bincount(y, minlength=len(LABELS))
    require(np.all(counts > 0), "Training fold is missing a label")
    w = (len(y) / (len(LABELS) * counts)) ** cfg["class_weight_power"]
    return torch.as_tensor(w, device=device, dtype=torch.float32)

# Restore sample order after length-based batching and convert logits to probabilities.
def predict(model, tokenizer, enc, batch_size, device, pad_multiple, y=None):
    started = time.perf_counter()
    model.eval()
    result = np.empty((len(enc), len(LABELS)), dtype=np.float32)
    with torch.inference_mode():
        for ids, tokens in batches(tokenizer, enc, batch_size, device, pad_multiple=pad_multiple):
            result[ids] = model(tokens).float().softmax(dim=-1).cpu().numpy()
    sync(device)
    loss = None if y is None else float(-np.log(np.maximum(result[np.arange(len(y)), y], 1e-30)).mean())
    require(np.isfinite(result).all(), "Invalid predicted probabilities")
    return result, loss, time.perf_counter() - started

def labels(rows):
    return np.array([LABELS.index(r["label"]) for r in rows], dtype=np.int64)

# Choose checkpoints on an inner grouped split, separate from the outer validation fold.
def split_fold(rows, fold, seed):
    outer_train = [r for r in rows if int(r["cv_fold"]) != fold]
    outer_valid = [r for r in rows if int(r["cv_fold"]) == fold]
    inner = StratifiedGroupKFold(n_splits=8, shuffle=True, random_state=seed + fold)
    fit, stop = next(inner.split(outer_train, labels(outer_train), [r["group_id"] for r in outer_train]))
    return [outer_train[i] for i in fit], [outer_train[i] for i in stop], outer_valid

def check_groups(*parts):
    accumulated = set()
    for rows in parts:
        groups = {r["group_id"] for r in rows}
        require(accumulated.isdisjoint(groups), "Train, early-stopping and evaluation groups overlap")
        accumulated |= groups

class EmotionClassifier(nn.Module):
    # Preserve the native pretrained dense/tanh/classification head.
    # Keep the frozen encoder in evaluation mode even during head training.
    def __init__(self, native, indices):
        super().__init__()
        self.encoder = native.roberta
        self.classifier = native.classifier
        self.encoder.requires_grad_(False)
        self.classifier.requires_grad_(True)
        self.register_buffer("emotion_indices", torch.tensor(indices, dtype=torch.long),
                             persistent=False)

    def train(self, mode=True):
        super().train(mode)
        self.encoder.eval()
        return self

    def forward(self, tokens):
        logits = self.classifier(self.encoder(**tokens).last_hidden_state)
        # Exclude the neutral logit before computing six-class probabilities.
        return logits.index_select(-1, self.emotion_indices)


def load_pretrained(cfg):
    if cfg.get("emotion_model"):
        local = MODEL_DIR / cfg["model"].split("/")[-1]
        path = local if local.exists() else cfg["model"]
        native = AutoModelForSequenceClassification.from_pretrained(
            path, trust_remote_code=False, attn_implementation="eager").float()
        mapping = {str(v).lower(): int(k) for k, v in native.config.id2label.items()}
        require(mapping == cfg["native_label2id"], "Native label mapping changed")
        if local.exists():
            require(sha(local / "config.json") == cfg["native_config_sha256"],
                    "Original native config changed")
        tokenizer = AutoTokenizer.from_pretrained(path, trust_remote_code=False)
        return tokenizer, EmotionClassifier(native, [mapping[label] for label in LABELS])

    # Load local pretrained weights or the pinned public model revision.
    local = MODEL_DIR / Path(cfg["local_pretrained"]).relative_to("pretrained")
    if local.exists():
        require(sha(local / "model.safetensors") == cfg["weight_sha256"],
                "Original pretrained weights changed")
        tokenizer = AutoTokenizer.from_pretrained(local, use_fast=True, trust_remote_code=False)
        encoder = AutoModel.from_pretrained(local, trust_remote_code=False,
                    use_safetensors=True, attn_implementation=cfg["attention"])
    else:
        params = dict(revision=cfg["revision"], trust_remote_code=False)
        tokenizer = AutoTokenizer.from_pretrained(cfg["model"], use_fast=True, **params)
        encoder_class = CLIPTextModel if cfg["pooling"] == "eos" else AutoModel
        encoder = encoder_class.from_pretrained(cfg["model"],
                    attn_implementation=cfg["attention"], **params)
    return tokenizer, Classifier(encoder, cfg["dropout"], cfg["pooling"])


def train_transformer(cfg, fit_rows, stop_rows, evaluation, device, seed, epochs):
    # Optimize cross-entropy and select the best checkpoint by inner Macro-F1.
    torch.set_num_threads(2)
    torch.manual_seed(seed)
    train_y = labels(fit_rows)
    stop_y = labels(stop_rows) if stop_rows else None
    tokenizer, model = load_pretrained(cfg)
    fit_enc = encode(tokenizer, fit_rows, cfg["max_length"])
    stop_enc = encode(tokenizer, stop_rows, cfg["max_length"]) if stop_rows else None
    if stop_rows:
        evaluation_enc = encode(tokenizer, evaluation, cfg["max_length"])
    model.to(device)
    batch_size = cfg["batch_size"]
    torch.manual_seed(seed)
    accumulation = max(1, math.ceil(cfg["effective_batch"] / batch_size))
    batch_count = math.ceil(len(fit_rows) / batch_size)
    per_epoch_steps = math.ceil(batch_count / accumulation)
    # Separate encoder/head learning rates and weight-decay groups.
    optimizer_groups = {}
    for name, parameter in model.named_parameters():
        decay = cfg["weight_decay"] if parameter.ndim > 1 and not name.endswith("bias") else 0.0
        lr = cfg["lr"] * cfg.get("head_lr_multiplier", 1.0) if name.startswith("classifier.") else cfg["lr"]
        optimizer_groups.setdefault((lr, decay), []).append(parameter)
    optimizer = torch.optim.AdamW(
        [{"params": params, "lr": lr, "weight_decay": decay}
         for (lr, decay), params in optimizer_groups.items()], lr=cfg["lr"])
    # Apply linear warmup and decay over the configured training schedule.
    schedule_steps = per_epoch_steps * cfg["epochs"]
    schedule = get_scheduler(cfg["scheduler"], optimizer=optimizer,
                 num_warmup_steps=int(schedule_steps * cfg["warmup"]),
                 num_training_steps=schedule_steps)
    weights_tensor = loss_weights(train_y, cfg, device)
    train_tensor = torch.as_tensor(train_y, device=device)
    best_f1, best_epoch, best_state = -1.0, 0, None
    compact = cfg.get("emotion_model", False)
    trainable = {name for name, p in model.named_parameters() if p.requires_grad}
    for epoch in range(1, epochs + 1):
        model.train()
        optimizer.zero_grad(set_to_none=True)
        for step, (ids, tokens) in enumerate(batches(
                tokenizer, fit_enc, batch_size, device, seed + epoch, cfg["pad_multiple"])):
            targets = train_tensor[torch.as_tensor(ids, device=device)]
            logits = model(tokens).float()
            loss = nn.functional.cross_entropy(logits, targets, weight=weights_tensor,
                                               label_smoothing=cfg.get("label_smoothing", 0.0))
            # Accumulate gradients to reach the effective batch size.
            divisor = min(accumulation, batch_count - step // accumulation * accumulation)
            (loss / divisor).backward()
            if (step + 1) % accumulation == 0 or step + 1 == batch_count:
                nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()
                schedule.step()
                optimizer.zero_grad(set_to_none=True)
        sync(device)
        score = None
        if stop_rows:
            inner_probs = predict(model, tokenizer, stop_enc, batch_size, device,
                                  cfg["pad_multiple"], stop_y)[0]
            score = float(f1_score(stop_y, inner_probs.argmax(axis=1),
                         labels=range(len(LABELS)), average="macro", zero_division=0))
        # Retain the best inner-validation state; head-only models need only trainable weights.
        if not stop_rows or score > best_f1:
            best_f1, best_epoch = score, epoch
            best_state = {k: v.detach().cpu().clone() for k,v in model.state_dict().items()
                          if not compact or k in trainable}
        # Stop after the minimum training duration if Macro-F1 stops improving.
        if stop_rows and epoch >= cfg.get("min_epochs", 1) and epoch - best_epoch >= cfg["patience"]:
            break
    missing, unexpected = model.load_state_dict(best_state, strict=not compact)
    if compact:
        require(not unexpected and not (set(missing) & trainable), "Incomplete head")
    del best_state
    if not stop_rows:
        evaluation_enc = encode(tokenizer, evaluation, cfg["max_length"])
    probs = predict(model, tokenizer, evaluation_enc, batch_size, device, cfg["pad_multiple"])[0]
    inner_probs = (predict(model, tokenizer, stop_enc, batch_size, device,
                          cfg["pad_multiple"], stop_y)[0] if stop_rows else None)
    return probs, best_epoch, inner_probs


def train_classical(training, test):
    # Fit a class-balanced linear SVM on the selected sparse features.
    with threadpool_limits(limits=1):
        X, V = classical_features(training, test)
        model = LinearSVC(C=.3, class_weight="balanced", penalty="l2",
                          loss="squared_hinge", dual=True, random_state=42, max_iter=10000)
        model.fit(X, [r["label"] for r in training])
        return model.predict(V)


# Fixed epoch counts selected during cross-validation for full-data E08 training.
FINAL_EPOCHS = {'T01_minilm_mean_lr5e5': 3,
 'T03_minilm_mean_ce256': 4,
 'T11_minilm_seed52': 4,
 'T02_clip_text_eos': 3,
 'T10_clip_head_lr5x': 3}


def train_e08(training, test):
    members = {}
    for name, cfg in TRANSFORMER_CONFIGS.items():
        # Train each ensemble member on the complete labeled dataset.
        members[name], _, _ = train_transformer(
            cfg, training, [], test, torch.device("cpu"), cfg["seed"], FINAL_EPOCHS[name])
        gc.collect()
    return members


# Class-specific log-probability offsets fitted on grouped inner-validation predictions.
E12_BIAS = np.array([-0.04111387509379578, 0.1006043016951198, 0.08831854071101616, -0.12403196972667603, -0.0411270767446843, 0.01735007915902018])

def fit_calibration(scores, truth):
    # Fit regularized class offsets using inverse-frequency-weighted log loss.
    logits = np.log(np.maximum(scores, 1e-12))
    counts = np.bincount(truth, minlength=len(LABELS))
    require(np.all(counts > 0), "Calibration needs all six labels")
    weights = counts[truth].astype(float) ** -.5
    weights /= weights.sum()
    index = np.arange(len(truth))
    def objective(bias):
        shifted = logits + bias
        log_norm = logsumexp(shifted, axis=1)
        loss = weights @ (log_norm - shifted[index, truth]) + .5 * .1 * (bias @ bias)
        probabilities = np.exp(shifted - log_norm[:, None])
        grad = weights @ probabilities
        grad -= np.bincount(truth, weights=weights, minlength=len(LABELS))
        return loss, grad + .1 * bias
    fit = minimize(objective, np.zeros(len(LABELS)), jac=True, method="L-BFGS-B")
    require(fit.success and np.isfinite(fit.x).all(), "Calibration failed")
    return fit.x


# Combine arithmetic, geometric and class-calibrated probability averages.
def aggregate_e08_e24(members):
    scores = {m: np.asarray(p, dtype=np.float64) for m, p in members.items()}
    e08 = np.sum([scores[m] * w for m,w in E08_MEMBERS.items()], axis=0)
    e18 = softmax(np.sum([np.log(np.maximum(scores[m], 1e-12)) * w
                          for m,w in E08_MEMBERS.items()], axis=0), axis=1)
    e12 = (.25 * scores["T01_minilm_mean_lr5e5"]
           + .25 * scores["T11_minilm_seed52"]
           + .50 * scores["T02_clip_text_eos"])
    e12 = softmax(np.log(np.maximum(e12, 1e-12)) + E12_BIAS, axis=1)
    return e08, .5 * e12 + .5 * e18


# Extract normalized frozen embeddings for the kernel and pairwise classifiers.
def extract_kernel_view(view, rows):
    member, variant = {"minilm": ("T01_minilm_mean_lr5e5", "raw"),
                       "clip": ("T02_clip_text_eos", "raw"),
                       "social": ("T01_minilm_mean_lr5e5", "social_names")}[view]
    cfg = TRANSFORMER_CONFIGS[member]
    device = torch.device("cpu")
    tokenizer, model = load_pretrained(cfg)
    net = model.encoder.to(device).eval()
    enc = encode(tokenizer, [{**r, "text": text_view(r["text"], variant)} for r in rows],
                 cfg["max_length"])
    result = np.empty((len(rows), net.config.hidden_size), dtype=np.float32)
    with torch.inference_mode():
        for ids, batch in batches(tokenizer, enc, 32, device, pad_multiple=8):
            outputs = net(**batch)
            if cfg["pooling"] == "eos":
                emission = outputs.pooler_output
            else:
                mask = batch["attention_mask"].unsqueeze(-1)
                emission = (outputs.last_hidden_state * mask).sum(dim=1) / mask.sum(dim=1).clamp_min(1)
            result[ids] = nn.functional.normalize(emission.float(), dim=-1).cpu().numpy()
    require(np.isfinite(result).all(), "Nonfinite embeddings")
    return result

KERNEL_CONFIG = {'method': 'weighted_kernel_ridge',
 'seed': 42,
 'ridge': 0.28,
 'frequency_power': 0.35,
 'alpha': 1.75,
 'temperature': 0.21,
 'vectorizers': {'word': {'analyzer': 'word',
                          'ngram_range': [1, 2],
                          'min_df': 2,
                          'sublinear_tf': True},
                 'char': {'analyzer': 'char',
                          'ngram_range': [3, 5],
                          'min_df': 2,
                          'sublinear_tf': True}},
 'weights': {'minilm': 0.33, 'social': 0.33, 'clip': 0.14, 'char': 0.15, 'word': 0.05}}

def lexical(cfg, rows, vectors):
    texts = [text_view(r["text"], "social_names") for r in rows]
    features = {}
    for key, vectorizer in vectors.items():
        features[key] = vectorizer.transform(texts)
    return features

# Sum nonlinear semantic similarities and linear TF-IDF similarities.
def kernel(train, other, cfg):
    shape = (next(iter(other.values())).shape[0], next(iter(train.values())).shape[0])
    result = np.zeros(shape, dtype=np.float32)
    for view, weight in cfg["weights"].items():
        product = other[view] @ train[view].T
        if view not in cfg["vectorizers"]:
            np.clip(product, -1, 1, out=product)
            product = np.exp(cfg["alpha"] * (product - 1))
        else:
            product = product.toarray()
        result += weight * product
    return result

def infer(model, rows, embeddings):
    cfg = model["cfg"]
    features = {**embeddings, **lexical(cfg, rows, model["vectorizers"])}
    return softmax((kernel(model["train"], features, cfg) @ model["coeff"]) / cfg["temperature"], axis=1)

def fit(cfg, rows, embeddings):
    texts = [text_view(r["text"], "social_names") for r in rows]
    vectorizers = {}
    features = dict(embeddings)
    for key, values in cfg["vectorizers"].items():
        if key not in cfg["weights"]:
            continue
        params = {**values, "ngram_range": tuple(values["ngram_range"])}
        vectorizers[key] = TfidfVectorizer(**params, dtype=np.float32)
        features[key] = vectorizers[key].fit_transform(texts)
    y = np.asarray([LABELS.index(r["label"]) for r in rows])
    class_count = np.bincount(y, minlength=len(LABELS))
    require((class_count > 0).all(), "Training omitted a label")
    sample_weight = (len(y) / (len(LABELS) * class_count[y])) ** cfg["frequency_power"]
    k = kernel(features, features, cfg)
    # Apply class-weighted ridge regularization before solving the linear system.
    k.flat[::len(y)+1] += cfg["ridge"] / sample_weight
    target = np.eye(len(LABELS), dtype=np.float32)[y]
    coeff = cho_solve(cho_factor(k, overwrite_a=True, check_finite=False), target, check_finite=False)
    return {"cfg": cfg, "vectorizers": vectorizers, "train": features, "coeff": coeff}

# Define binary emotion pairs and the four frequently confused pairs.
PAIRS = list(itertools.combinations(range(len(LABELS)), 2))
TARGET_PAIRS = (("anger", "disgust"), ("joy", "surprise"), ("joy", "sadness"), ("anger", "sadness"))
FOCUS = {tuple(sorted(map(LABELS.index, pair))) for pair in TARGET_PAIRS}
CONFIG = {
    "seed": 613,
    "model": {"solver": "liblinear", "C": 6., "regularization": "l2", "ratio_smoothing": 1.,
              "word_scale": 1., "char_scale": .8, "semantic_scale": 3.5, "class_balance_power": .25},
    "text": {"word": {"analyzer": "word", "ngram_range": [1, 2], "max_features": 32000},
             "char": {"analyzer": "char", "ngram_range": [3, 5], "max_features": 36000}},
    "decisions": {
        "P6E01_pair_experts": {"mix": .45, "gap": .22, "pairs": "all", "member": "hybrid"},
        "P6E02_confusion_experts": {"mix": .55, "gap": .22, "pairs": "common", "member": "hybrid"},
        "P6E03_lexical_experts": {"mix": .55, "gap": .16, "pairs": "common", "member": "lexical"},
        "P6E04_pair_consensus": {"mix": .65, "gap": .28, "pairs": "all", "member": "consensus"},
    },
    "promotion": {"min_gain": .0015, "min_wins": 3},
}

# Build binary word and character counts using training-only vocabularies.
def text_features(train, val):
    fit = {}
    transform = {}
    hold = {}
    for key, params in CONFIG["text"].items():
        arguments = {**params, "ngram_range": tuple(params["ngram_range"])}
        if params["analyzer"] == "word":
            arguments["token_pattern"] = r"(?u)\b[\w']+\b"
        vectorizer = CountVectorizer(**arguments, binary=True, min_df=2, dtype=np.float32)
        fit[key] = vectorizer.fit_transform([r["text"] for r in train])
        transform[key] = vectorizer.transform([r["text"] for r in val])
        hold[key] = vectorizer
    return fit, transform, hold

# Combine raw and social-text MiniLM embeddings with CLIP embeddings.
def semantic_features(views):
    raw = normalize(.75 * views["minilm"] + .25 * views["social"])
    both = np.column_stack([raw * .8, views["clip"] * .6])
    return sparse.csr_matrix(both * CONFIG["model"]["semantic_scale"])

# Weight lexical features by smoothed class-conditional log-count ratios.
def nb_features(train, val, index, target):
    lexical_train, lexical_val, ratios = [], [], {}
    for key, samples in train.items():
        data = samples[index]
        smoothing = CONFIG["model"]["ratio_smoothing"]
        p = np.asarray(data[target].sum(axis=0)).ravel() + smoothing
        q = np.asarray(data[~target].sum(axis=0)).ravel() + smoothing
        ratio = np.log((p / p.sum()) / (q / q.sum()))
        ratios[key] = ratio
        weight = CONFIG["model"][key + "_scale"]
        lexical_train.append(normalize(data).multiply(weight * ratio))
        lexical_val.append(normalize(val[key]).multiply(weight * ratio))
    return sparse.hstack(lexical_train, format="csr"), sparse.hstack(lexical_val, format="csr"), ratios

# Train binary logistic-regression specialists using only each pair's training examples.
def pair_fit(training, validation, z_train, z_val):
    train_text, val_text, vectors = text_features(training, validation)
    train_sem, val_sem = semantic_features(z_train), semantic_features(z_val)
    y = np.asarray([LABELS.index(r["label"]) for r in training])
    coefficients = {}
    scores = {"hybrid": {}, "lexical": {}}
    for a, b in PAIRS:
        index = np.where((y == a) | (y == b))[0]
        target = y[index] == b
        class_counts = np.bincount(target.astype(int))
        weights = (len(target) / (2 * class_counts[target.astype(int)])) ** CONFIG["model"]["class_balance_power"]
        left, right, ratio = nb_features(train_text, val_text, index, target)
        models = {}
        features = {"hybrid": (sparse.hstack([left, train_sem[index]], format="csr"),
                               sparse.hstack([right, val_sem], format="csr")),
                    "lexical": (left, right)}
        for name, (x, val) in features.items():
            solver = LogisticRegression(C=CONFIG["model"]["C"], solver=CONFIG["model"]["solver"],
                                        random_state=CONFIG["seed"], max_iter=150)
            solver.fit(x, target, sample_weight=weights)
            models[name] = solver
            scores[name][a, b] = solver.predict_proba(val)[:, 1]
        coefficients[a, b] = {"ratios": ratio, "models": models}
    return scores, {"vectorizers": vectors, "pairs": coefficients}

def adjust(base, specialists, config):
    result = base.copy()
    top_two = np.argsort(base, axis=1)[:, -2:]
    nearest = np.sort(top_two, axis=1)
    top_score = np.sort(base, axis=1)[:, -2:]
    # Activate specialists only for uncertain predictions between the relevant top two labels.
    weak = top_score[:, 1] - top_score[:, 0] < config["gap"]
    for a, b in PAIRS:
        if config["pairs"] == "common" and (a, b) not in FOCUS:
            continue
        eligible = weak & (nearest[:, 0] == a) & (nearest[:, 1] == b)
        member = config["member"]
        if member == "consensus":
            hybrid = specialists["hybrid"][a, b]
            lexical = specialists["lexical"][a, b]
            eligible &= (hybrid > .5) == (lexical > .5)
            comparison = .65 * hybrid + .35 * lexical
        else:
            comparison = specialists[member][a, b]
        total = base[eligible, a] + base[eligible, b]
        prior = base[eligible, b] / total
        w = config["mix"]
        # Mix conditional log-odds while preserving the pair's total probability mass.
        posterior = expit((1 - w) * logit(np.clip(prior, 1e-8, 1 - 1e-8))
                          + w * logit(np.clip(comparison[eligible], 1e-5, 1 - 1e-5)))
        result[eligible, b] = total * posterior
        result[eligible, a] = total * (1 - posterior)
    require(np.isfinite(result).all() and np.allclose(result.sum(1), 1, atol=1e-5), "Invalid expert score")
    return result


def train_p6(training, test, e24):
    views = {name: extract_kernel_view(name, training + test)
             for name in ("minilm", "clip", "social")}
    n = len(training)
    z_train = {k: v[:n] for k,v in views.items()}
    z_test = {k: v[n:] for k,v in views.items()}
    kernel_model = fit(KERNEL_CONFIG, training, z_train)
    kernel_probs = infer(kernel_model, test, z_test)
    # Assign the kernel 35% weight for low-margin predictions and 5% otherwise.
    ordered = np.sort(e24, axis=1)
    gap = ordered[:, -1] - ordered[:, -2]
    weight = np.where(gap < .20, .35, .05)[:, None]
    p5 = (1 - weight) * e24 + weight * kernel_probs
    specialist_probs, _ = pair_fit(training, test, z_train, z_test)
    return adjust(p5, specialist_probs, CONFIG["decisions"]["P6E02_confusion_experts"])


def train_emotion_cv_heads(development, test, cfg):
    # Average test probabilities from five heads selected on grouped inner validation.
    device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
    all_probs = []
    for fold in range(5):
        fit_rows, stop_rows, outer = split_fold(development, fold, cfg["seed"])
        check_groups(fit_rows, stop_rows, outer)
        # Select on the inner validation only. Predict test with the selected head.
        probs, epoch, _ = train_transformer(cfg, fit_rows, stop_rows, test, device,
                                           cfg["seed"] + fold, cfg["epochs"])
        all_probs.append(probs)
        print(cfg["id"], "fold", fold, "selected epoch", epoch)
        gc.collect()
        if device.type == "mps":
            torch.mps.empty_cache()
    return np.mean(all_probs, axis=0)


def final_submission(training, development, test):
    member_probs = train_e08(training, test)
    e08, e24 = aggregate_e08_e24(member_probs)
    p6 = train_p6(training, test, e24)
    em06 = train_emotion_cv_heads(development, test, EMOTION_CONFIGS["EM06_large_head"])
    em03 = train_emotion_cv_heads(development, test, EMOTION_CONFIGS["EM03_distil_head"])
    # Combine EM06, E08, P6E02 and EM03 with weights 48%, 16%, 16% and 20%.
    triple = .60 * em06 + .20 * e08 + .20 * p6
    scores = .80 * triple + .20 * em03
    require(scores.shape == (1440, 6), "Wrong score dimensions")
    require(np.isfinite(scores).all() and np.allclose(scores.sum(1), 1, atol=1e-4),
            "Invalid scores")
    return pd.DataFrame({"id": [r["post_id"] for r in test],
                         "label": np.asarray(LABELS)[scores.argmax(1)]})


# Enable training and CSV export explicitly; requires competition data and pretrained weights.
RUN_TRAINING = False
if RUN_TRAINING:
    corpus, training, test = load_competition()
    development = frozen_development(corpus)
    # Export classical baseline predictions in Kaggle format.
    classical = pd.DataFrame({"id": [r["post_id"] for r in test],
                              "label": train_classical(training, test)})
    classical.to_csv("C1_structure_retrained.csv", index=False)
    # Export the final ensemble's predicted emotion labels.
    submission = final_submission(training, development, test)
    require(submission.id.is_unique and submission.label.isin(LABELS).all(), "Invalid submission")
    submission.to_csv("DM2026_final_retrained.csv", index=False)
    print(submission.head())
```
